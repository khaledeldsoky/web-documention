How upload the RPMs on NEXUS

```python
for rpm in *.rpm; do
  echo "Uploading $rpm..."
  curl -v -u admin:root@123 --upload-file "$rpm" http://nexus.apps.local/repository/yum-repo/"$rpm"
done
```
